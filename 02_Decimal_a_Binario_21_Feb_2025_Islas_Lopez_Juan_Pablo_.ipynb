# I) La conversión de un número entero en base 10 al sistema binario

#Una forma sencilla de convertir decimal en binario en Python es dividir sucesivamente el número decimal entre 2 e ir guardando el módulo, resto o residuo de cada división hasta obtener cero en el cociente. Finalmente se concatenan todos los módulos en orden inverso para obtener el número en binario.

#Paso 1

#Se trata de dividir el número de origen entre 2. Además, necesitaremos dos números distintos resultado de esta división: el cociente y el módulo
#Al hacer una división entera con la doble barra obtendrás el cociente.
# También necesitamos obtener el módulo, para ello tendremos que aplicar el operador %, que se encarga de resolver la división entera nuevamente, pero esta vez el resultado es el módulo.

# Paso 2
#Guardar el módulo de la división entre 2
# En el segundo paso tienes que guardar el módulo obtenido en la división previa. Como este es un proceso iterativo donde vamos a guardar diversos módulos lo mejor es crear una lista e ir almacenándolos en ella.

#Paso 3
#Repetir los pasos 1 y 2 hasta obtener 0 en el cociente
#Este es un proceso iterativo, así que necesitamos realizar los pasos 1 y 2 en un bucle teniendo en cuenta dos aspectos principales:

#Cada vez que obtenemos un cociente, este pasa a ser el número que utilizamos en la siguiente iteración para realizar la división.
#Debemos detener el bucle cuando el cociente obtenido sea 0.

```python
numero_decimal = 26 # este es el número que queremos convertir a binario

modulos = [] # la lista para guardar los módulos

while numero_decimal != 0: # mientras el número de entrada sea diferente de cero
    # paso 1: dividimos entre 2
    modulo = numero_decimal % 2
    cociente = numero_decimal // 2
    modulos.append(modulo) # guardamos el módulo calculado
    numero_decimal = cociente # el cociente pasa a ser el número de entrada

print(modulos)
```

#Paso 4: Invertir los módulos guardados
#Una vez obtenidos los módulos hay que darles la vuelta, pues el número binario resultante se compone de todos estos módulos concatenados pero en orden inverso al de obtención. Esto se puede realizar muy fácil utilizando la notación de porciones para obtener toda la lista al revés de la siguiente manera: modulos = modulos[::-1].
#Para hacer esto podemos hacer uso de la función insert que nos permite introducir un elemento en una lista pero en la posición que queramos.


```python
numero_decimal = 26

modulos = []
while numero_decimal != 0:
    # se almacena el módulo en el orden correcto
    modulos.insert(0, numero_decimal % 2)
    numero_decimal //= 2

print(modulos)
```

# III)La conversión de un número decimal en base 10 al sistema binario.

#Usaremos los mismos 5 pasos que vimos cuando se pasaba de un numero  entero en base 10 a un binario, solo que la diferencia es que el la parte decimal se utiliza un algoritmo distinto  para pasar la parte decimal a binaria el cual consiste en  multiplicar la parte decimal por 2 y despues  separar la parte entera y la decimal restante seguirla multiplicando por 2 y repetir estas operaciones hasta que la parte decimal restante se reduzca a cero

```python
numero_decimal = 12.75
 # Se me hizo mas facil separa la parte entera y la parte decimal del numero de base 10

parte_entera = int(numero_decimal)
parte_decimal = abs(numero_decimal) - abs(int(numero_decimal))
# Ya con la parte entera separada y renombrandola , se aplican los mis 5 pasos que el apartado I
modulos = []
modulos_parte_decimal = []
while parte_entera != 0:
    # se almacena el módulo en el orden correcto
    modulos.insert(0, parte_entera % 2)
    parte_entera //= 2
# De la manera mas rudimentaria , literal lo hice paso por paso para pasar este numero en especifico 12.75 a binario , y la parte decimal ,
# primero la multiplique por 2 osea 0.75(*2) = 1.5 , quite la parte entera  1.5 -1.0 =0.5 y nuevamente la parte decimal la multiplique por 2 "2*0.5)=1.0"
#y le quite la parte entera "1.0-1=0.0 hasta que la parte decimal se redujo a cero y en compracion con la parte entera , aqui el orde no los "1 y 0" no se cambian
modulos_parte_decimal.insert(0,int(parte_decimal*2))
modulos_parte_decimal.insert(1,abs(int(parte_decimal*2))-abs(int(int(parte_decimal))))

print (parte_entera)
print(parte_decimal)
print(modulos)
print(modulos_parte_decimal) #Por ultimo 12.75 a binario es 1100.11"
```
